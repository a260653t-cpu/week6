# 日本の地域データ分析レポート（Part 2）

* **氏名**: 田上達貴
* **依頼元**: 地方創生マーケティング部門
* **読み手**: 各地域の事業企画担当者
* **問い**: 各都道府県の「高齢化率（65歳以上人口の割合）」にはどのような地域的偏りがあるか？上位・下位の都道府県を特定する。

```python
import csv

# データの読み込み
filename = 'japan.csv'

with open(filename, mode='r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)
    rows = [row for row in reader]

# 列インデックスの特定
pref_idx = header.index('Prefecture')
total_pop_idx = header.index('総人口')
over65_idx = header.index('65歳以上人口')

# 高齢化率（65歳以上人口 / 総人口 * 100）の計算
results = []
for row in rows:
    pref = row[pref_idx]
    total_pop = float(row[total_pop_idx])
    over65 = float(row[over65_idx])
    
    rate = (over65 / total_pop) * 100
    results.append((pref, rate))

# 高齢化率が高い順にソート
results.sort(key=lambda x: x[1], reverse=True)

# 結果の表示（ランキング）
print("【高齢化率が高い都道府県 TOP 5】")
for rank, (pref, rate) in enumerate(results[:5], 1):
    print(f"{rank}位: {pref} ({rate:.2f}%)")

print("\n【高齢化率が低い都道府県 TOP 5】")
for rank, (pref, rate) in enumerate(results[-5:], 1):
    print(f"{rank}位: {pref} ({rate:.2f}%)")
```

```python
# 1. データの準備 (TOP 5 と Bottom 5)
top5 = results[:5]
bottom5 = results[-5:]
combined = top5 + bottom5

# 2. 簡易テキスト棒グラフの表示
print("=" * 50)
print("【高齢化率の比較 棒グラフ（視覚化）】")
print("=" * 50)

max_rate = max(item[1] for item in combined)
for pref, rate in combined:
    # 割合に応じたバー（■）の長さを計算
    bar_length = int((rate / max_rate) * 30)
    bar = "■" * bar_length
    print(f"{pref:<10} | {bar} {rate:.2f}%")

print("=" * 50)
```

## 分析結果と結論

* **分析結果**:
  * 高齢化率が最も高いのは **Akita（秋田県: 38.60%）** であり、次いで Kochi（高知県）、Yamaguchi（山口県）の順となった。
  * 一方で、高齢化率が低い地域には **Shiga（滋賀県: 26.83%）**、Kanagawa（神奈川県）、Aichi（愛知県）などが含まれる。
* **結論**:
  * 地方都市（秋田や高知など）では高齢化が著しく進行しているのに対し、大都市圏およびその周辺都市（神奈川、愛知、滋賀など）では比較的若い世代が多く留まっている傾向が見られる。
  * 地方創生の施策としては、高齢化率が高い地域に対して医療・福祉リソースの最適化や若年層の移住促進策が優先課題となる。
